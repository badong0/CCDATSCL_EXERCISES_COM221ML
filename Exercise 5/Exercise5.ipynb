## Part 1: ANOVA

A teacher wants to investigate whether different teaching methods affect students’ exam scores. The dataset contains 50 students, each assigned to one of three teaching methods (A, B, or C).

Question: "Is there a difference in mean exam scores among students taught using three different teaching methods?"

- Independent variable (factor): Teaching Method (A, B, C)
- Dependent variable: Exam Score (numeric)

```python
url = "https://github.com/robitussin/CCDATSCL_EXERCISES/blob/f56f00991116fcb717a8827d324aa6f599dea366/exam_scores.csv?raw=true"

df_scores = pd.read_csv(url)
df_scores.head()
```

Calculate the number of students in each teaching method.

```python
counts_per_method = df_scores['Method'].value_counts()
print("Student distribution per method:")
print(counts_per_method)
```

Compute the mean and standard deviation of exam scores for each group.

```python
# Using aggregation for a cleaner output
score_stats = df_scores.groupby('Method')['Score'].agg(['mean', 'std'])
print("Descriptive statistics for each group:")
print(score_stats)
```

Use a plot to show the exam scores by teaching method.

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Method', y='Score', data=df_scores, hue='Method', palette='viridis', legend=False)
plt.title('Analysis of Exam Scores by Methodology')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

Perform the ANOVA Test
- Use scipy.stats.f_oneway to conduct a one-way ANOVA.
- Report the F-statistic and p-value.
- Using a significance level of α = 0.05, state whether you reject or fail to reject H₀.

```python
data_a = df_scores[df_scores['Method'] == 'A']['Score']
data_b = df_scores[df_scores['Method'] == 'B']['Score']
data_c = df_scores[df_scores['Method'] == 'C']['Score']

f_stat, p_result = stats.f_oneway(data_a, data_b, data_c)

print(f"ANOVA F-statistic: {f_stat:.4f}")
print(f"ANOVA P-value: {p_result:.4f}")

sig_level = 0.05
if p_result < sig_level:
    print(f"Conclusion: Since {p_result:.4f} < {sig_level}, we reject the null hypothesis (H0).")
else:
    print(f"Conclusion: Since {p_result:.4f} > {sig_level}, we fail to reject the null hypothesis (H0).")
```

Interpret the results. What does it say about the relationship between teaching method and exam scores?

**INTERPRETATION**:
Batay sa resulta ng ANOVA, ang nakuha nating p-value (0.0000) ay higit na mababa sa 0.05, kaya nararapat lamang na i-reject ang null hypothesis. Ipinapakita nito na mayroong makabuluhang pagkakaiba sa mean scores ng mga estudyante depende sa teaching method na ginamit. Sa madaling salita, ang paraan ng pagtuturo ay may direktang impluwensya sa performance ng mga estudyante.

## Part 2: Chi-squared

A survey was conducted to see if there is a relationship between gender and littering behavior.

Question:
"Is gender associated with littering behavior?"

Each row represents one person.

```python
url_litter = "https://github.com/robitussin/CCDATSCL_EXERCISES/blob/f56f00991116fcb717a8827d324aa6f599dea366/litteringdata.csv?raw=true"

df_litter = pd.read_csv(url_litter)
df_litter.head()
```

Count the number of observations in each category for Gender and Litter separately.

```python
gender_freq = df_litter['Gender'].value_counts()
litter_freq = df_litter['Litter'].value_counts()

print("Gender frequency count:")
print(gender_freq)
print("\nLittering behavior frequency count:")
print(litter_freq)
```

Using `pd.crosstab()`, create a contingency table showing counts of littering behavior for each gender.
- Write down the row totals, column totals, and grand total from the table.

```python
xtab = pd.crosstab(df_litter['Gender'], df_litter['Litter'])
print("Observed Contingency Table:")
print(xtab)

sum_rows = xtab.sum(axis=1)
sum_cols = xtab.sum(axis=0)
total_all = xtab.values.sum()

print(f"\nRow Totals:\n{sum_rows}")
print(f"\nColumn Totals:\n{sum_cols}")
print(f"\nGrand Total: {total_all}")
```

State the null hypothesis (H₀) and alternative hypothesis (H₁) for this chi-squared test.

**Null Hypothesis (H₀)**: There is no association between gender and littering behavior (they are independent).

**Alternative Hypothesis (H₁)**: There is a significant association between gender and littering behavior.

Use scipy.stats.chi2_contingency to calculate:
- Chi-squared statistic
- p-value
- Degrees of freedom

Using a significance level of `α = 0.05`, determine whether to reject or fail to reject H₀.


```python
c_stat, p_val_chi, deg_free, exp_freq = chi2_contingency(xtab)

print(f"Chi-squared: {c_stat:.4f}")
print(f"P-value: {p_val_chi:.4f}")
print(f"DF: {deg_free}")

alpha_chi = 0.05
if p_val_chi < alpha_chi:
    print(f"\nResult: Reject H0 at alpha {alpha_chi}.")
else:
    print(f"\nResult: Fail to reject H0 at alpha {alpha_chi}.")
```

Interpret your results. Is there evidence that gender and littering behavior are associated?

**INTERPRETATION:** Dahil ang p-value (0.1120) ay mas mataas kaysa sa ating significance level (0.05), hindi tayo pwedeng mag-reject ng null hypothesis. Ibig sabihin, wala tayong sapat na ebidensya para sabihing magkaugnay ang kasarian at ang gawi sa pagkakalat sa dataset na ito.
